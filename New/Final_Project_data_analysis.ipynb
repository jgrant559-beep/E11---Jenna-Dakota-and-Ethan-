```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
all_data = {
    "radiation": {
        "dwt_berkeley0": pd.read_csv("radiation.csv"),
        "dwt_berkeley1": pd.read_csv("radiation(5).csv"),
        "north_berkeley0": pd.read_csv("radiation(1).csv"),
        "north_berkeley1": pd.read_csv("radiation(6).csv"),
        "macarthur0": pd.read_csv("radiation(2).csv"),
        "macarthur1": pd.read_csv("radiation(7).csv"),
        "west_oakland0": pd.read_csv("radiation(3).csv"),
        "west_oakland1": pd.read_csv("radiation(8).csv"),
        "embarcadero0": pd.read_csv("radiation(4).csv"),
        "embarcadero1": pd.read_csv("radiation(9).csv")
    },
    "voc": {
        "dwt_berkeley0": pd.read_csv("pth.csv").loc[:, "voc_mean":"voc_err"],
        "dwt_berkeley1": pd.read_csv("pth(5).csv").loc[:, "voc_mean":"voc_err"],
        "north_berkeley0": pd.read_csv("pth(1).csv").loc[:, "voc_mean":"voc_err"],
        "north_berkeley1": pd.read_csv("pth(6).csv").loc[:, "voc_mean":"voc_err"],
        "macarthur0": pd.read_csv("pth(2).csv").loc[:, "voc_mean":"voc_err"],
        "macarthur1": pd.read_csv("pth(7).csv").loc[:, "voc_mean":"voc_err"],
        "west_oakland0": pd.read_csv("pth(3).csv").loc[:, "voc_mean":"voc_err"],
        "west_oakland1": pd.read_csv("pth(8).csv").loc[:, "voc_mean":"voc_err"],
        "embarcadero0": pd.read_csv("pth(4).csv").loc[:, "voc_mean":"voc_err"],
        "embarcadero1": pd.read_csv("pth(9).csv").loc[:, "voc_mean":"voc_err"]
    },
    "humidity": {
        "dwt_berkeley0": pd.read_csv("pth.csv").loc[:, "hum_mean":"hum_err"],
        "dwt_berkeley1": pd.read_csv("pth(5).csv").loc[:, "hum_mean":"hum_err"],
        "north_berkeley0": pd.read_csv("pth(1).csv").loc[:, "hum_mean":"hum_err"],
        "north_berkeley1": pd.read_csv("pth(6).csv").loc[:, "hum_mean":"hum_err"],
        "macarthur0": pd.read_csv("pth(2).csv").loc[:, "hum_mean":"hum_err"],
        "macarthur1": pd.read_csv("pth(7).csv").loc[:, "hum_mean":"hum_err"],
        "west_oakland0": pd.read_csv("pth(3).csv").loc[:, "hum_mean":"hum_err"],
        "west_oakland1": pd.read_csv("pth(8).csv").loc[:, "hum_mean":"hum_err"],
        "embarcadero0": pd.read_csv("pth(4).csv").loc[:, "hum_mean":"hum_err"],
        "embarcadero1": pd.read_csv("pth(9).csv").loc[:, "hum_mean":"hum_err"]
    }
}
```

```python
print(f"Day 1:")
for key, value in all_data.get("radiation").items():
    data_size = len(value.loc[:, "cps"])
    xs = np.arange(data_size)*5 # 5 second intervals
    if str(key).endswith("0"):
        plt.plot(xs, value.loc[:, "cps"], label=key)
        mean_count = np.mean(value.loc[:, "cps"])
        sig_counts = np.sqrt(mean_count)
        mean_cps = mean_count / 5
        sig_cps = sig_counts / 5
        print(f"\t{key[:-1]} - Mean CPS = {mean_cps:.3f} +/- {sig_cps:.3f}")
plt.legend()
plt.tight_layout()
plt.title("Day 1 Counts At Different BART Stations")
plt.xlabel("Seconds")
plt.ylabel("Counts")
plt.show()
for key, value in all_data.get("radiation").items():
    data_size = len(value.loc[:, "usv_hr"])
    xs = np.arange(data_size)*5 # 5 second intervals
    if str(key).endswith("0"):
        plt.plot(xs, value.loc[:, "usv_hr"], label=key)
        mean_dose = np.mean(value.loc[:, "usv_hr"])
        sig_dose = np.std(value.loc[:, "usv_hr"])
        print(f"\t{key[:-1]} - Mean Dose = {mean_dose:.3f} +/- {sig_dose:.3f}")
plt.legend()
plt.tight_layout()
plt.title("Day 1 Dose At Different BART Stations")
plt.xlabel("Seconds")
plt.ylabel("usv per hour")
plt.show()
```

```python
print(f"Day 2:")
for key, value in all_data.get("radiation").items():
    data_size = len(value.loc[:, "cps"])
    xs = np.arange(data_size)*5 # 5 second intervals
    if str(key).endswith("1"):
        plt.plot(xs, value.loc[:, "cps"], label=key)
        mean_count = np.mean(value.loc[:, "cps"])
        sig_counts = np.sqrt(mean_count)
        mean_cps = mean_count / 5
        sig_cps = sig_counts / 5
        print(f"\t{key[:-1]} - Mean CPS = {mean_cps:.3f} +/- {sig_cps:.3f}")
plt.legend()
plt.tight_layout()
plt.title("Day 2 Counts At Different BART Stations")
plt.xlabel("Seconds")
plt.ylabel("Counts")
plt.show()
for key, value in all_data.get("radiation").items():
    data_size = len(value.loc[:, "usv_hr"])
    xs = np.arange(data_size)*5 # 5 second intervals
    if str(key).endswith("1"):
        plt.plot(xs, value.loc[:, "usv_hr"], label=key)
        mean_dose = np.mean(value.loc[:, "usv_hr"])
        sig_dose = np.std(value.loc[:, "usv_hr"])
        print(f"\t{key[:-1]} - Mean Dose = {mean_dose:.3f} +/- {sig_dose:.3f}")
plt.legend()
plt.tight_layout()
plt.title("Day 2 Dose At Different BART Stations")
plt.xlabel("Seconds")
plt.ylabel("usv per hour")
plt.show()
```

```python
trunc = 15
for key, value in all_data.get("voc").items():
    data_size = len(value.loc[:, "voc_mean"]) - trunc
    xs = np.arange(data_size)*5 # 5 second intervals
    if str(key).endswith("0"):
        plt.plot(xs, value.loc[trunc:, "voc_mean"], label=key)
        mean_voc = np.mean(value.loc[trunc:, "voc_mean"])
        sig_voc = np.std(value.loc[trunc:, "voc_mean"])
        print(f"\t{key[:-1]} - Mean VOC = {int(np.round(mean_voc))} +/- {int(np.round(sig_voc))}")
plt.legend()
plt.tight_layout()
plt.title("Day 1 Volatile Organic Compounds At Different BART Stations")
plt.xlabel("Seconds")
# plt.yscale("log")
plt.ylabel("Ohms")
plt.show()
for key, value in all_data.get("voc").items():
    data_size = len(value.loc[:, "voc_mean"]) - trunc
    xs = np.arange(data_size)*5 # 5 second intervals
    if str(key).endswith("1"):
        plt.plot(xs, value.loc[trunc:, "voc_mean"], label=key)
        mean_voc = np.mean(value.loc[trunc:, "voc_mean"])
        sig_voc = np.std(value.loc[trunc:, "voc_mean"])
        print(f"\t{key[:-1]} - Mean VOC = {int(np.round(mean_voc))} +/- {int(np.round(sig_voc))}")
plt.legend()
plt.tight_layout()
plt.title("Day 2 Volatile Organic Compounds At Different BART Stations")
plt.xlabel("Seconds")
plt.ylabel("Ohms")
plt.show()
```
